    if remainder is 0 for 3 and 5:

````python
#2 Python program that outputs a game of fizz buzz from 1 to 100

for i in range(1, 101):
    if i % 15 == 0:
        print("fizzbuzz")
    elif i % 3 == 0:
        print("fizz")
    elif i % 5 == 0:
        print("buzz")
    else:
        print(i)
````

````python
 # Calculate GC content from fasta

with open("Turkey_transcripts_15.fasta", "r") as infile, open("gc_content.txt", "w") as outfile:
    for line in infile:
        line = line.rstrip()
        gene_id, seq = line.split()
        gc = (seq.count("G") + seq.count("C")) / len(seq)
        outfile.write(gene_id + "\t" + str(gc) + "\n")
````

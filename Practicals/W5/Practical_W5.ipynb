```
#1 Pseudocode 

for numbers in range 1 through 100: 
    find the remainder when divided by 3 
    find the remainder when divided by 5 

    if remainder is 0 for 3 and 5: 
        print fizzbuzz
    elif remainder is 0 for 3: 
        print fizz
    elif remainder is 0 for 5:
        print buzz 
    else:

```
        
        



    
        

````python
#2 Python program that outputs a game of fizz buzz from 1 to 100

for i in range(1, 101):
    if i % 15 == 0:
        print("fizzbuzz")
    elif i % 3 == 0:
        print("fizz")
    elif i % 5 == 0:
        print("buzz")
    else:
        print(i)
````

```python
import pickle

genetic_code = pickle.load(open("genetic_code.pickle", "rb"))

def get_amino_acids(mRNA):
    i = 0
    aa_sequence = []
    while (i + 3) < len(mRNA):
        codon = mRNA[i:(i + 3)]
        aa = genetic_code[codon]
        if aa == "Stop":
            break
        else:
            aa_sequence.append(aa)
        i = i + 3
    return "".join(aa_sequence)
```

```python
with open("DataFiles/Turkey_transcripts_15_coding.fasta", "r") as f:
    header = f.readline().rstrip()
    seq = f.readline().rstrip()

seq = seq.replace("T", "U")
protein = get_amino_acids(seq)

print(protein)
```

```python
with open("DataFiles/Turkey_transcripts_15_coding.fasta", "r") as infile, open("Turkey_proteins.fasta", "w") as outfile:
    for line in infile:
        line = line.rstrip()

        if line[0] == ">":
            header = line.replace("gbkey=CDS", "gbkey=AA")
            outfile.write(header + "\n")

        else:
            seq = line.replace("T", "U")
            protein = get_amino_acids(seq)
            outfile.write(protein + "\n")
```
